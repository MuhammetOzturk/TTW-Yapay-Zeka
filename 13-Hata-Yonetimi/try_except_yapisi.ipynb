```python
try:
    a = int(input('1. sayiyi giriniz:'))
    b = int(input('2. sayiyi giriniz:'))
    a / b
except ValueError:
    print("Hatali veri girisi yapildi.")
except ZeroDivisionError:
    print('Sifira bolme hatasi olustu.')

print('Program sonlandi.')
```

```python
try:
    sayi = float(input("Bir sayi giriniz:"))
    print(sayi + 10)
except TypeError:
    print("Hata olustu")
except ValueError:
    print("Hatali veri girisi!")
```

```python
try:
    sayi = int(input("Bir sayi giriniz:"))
    print(sayi + 10)
except Exception as hata:
    print(hata)
print("Program bitti.")
```

```python
def bol(a,b):
    try:
        return a/b
    except ZeroDivisionError:
        print("0'a bolme hatasi olustu.")

bol(2,0)
bol(2,0)
```
