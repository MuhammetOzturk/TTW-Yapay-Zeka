### Soru 2: Geri Sayım

10'dan 1'e kadar geri sayım yapan ve her sayıyı ekrana yazdıran, ardından 'Başlat!' mesajını gösteren bir `while` döngüsü yazın.

````python
import time
sayac = 10
# Kodunuzu buraya yazın

while sayac > 0:
    print(sayac)
    sayac -= 1
    time.sleep(1)

print("Başlat!")
````


### Soru 5: Yıldız Deseni Oluşturma

İç içe `for` döngüleri kullanarak aşağıdaki gibi bir yıldız deseni oluşturan bir kod yazın:

```
*
**
***
****
*****
```

````python
# Kodunuzu buraya yazın
for i in range(1, 6):
    print("*" * i)
````
