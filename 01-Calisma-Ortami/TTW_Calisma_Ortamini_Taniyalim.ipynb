
# JUPYTER NOTEBOOK KLAVYE KISAYOLLARI

Jupyter’da iki mod var:
**Command Mode (mavi çerçeve)**
**Edit Mode (yeşil çerçeve)**

## 1. Modlar Arası Geçiş

* Enter: Edit Mode
* Esc: Command Mode

## 2. Hücre İşlemleri (Command Mode)

* A: Üste hücre ekle
* B: Alta hücre ekle
* DD: Hücre sil
* M: Hücreyi Markdown yap
* Y: Hücreyi Code yap
* R: Hücreyi Raw yap
* Shift + M: Hücreleri birleştir
* Ctrl + Shift + - : Hücreyi böl
* O: Hücre output’unu gizle/göster
* Shift + O: Hücre output’unu satır kaydırmalı yap
* L: Satır numaralarını aç/kapat
* S: Kaydet
* Z: Son hücre silinmesini geri al

## 3. Çalıştırma & Kernel

* Shift + Enter: Çalıştır, aşağı geç
* Ctrl + Enter: Çalıştır, aynı hücrede kal
* Alt + Enter: Çalıştır, yeni hücre aç
* 0 0 (iki kere 0): Kernel restart
* I I (iki kere I): Kernel interrupt

## 4. Navigasyon

* Up / Down: Hücreler arası geçiş
* K: Yukarı git
* J: Aşağı git

## 5. Arama, Menü, Yardım

* F: Arama
* H: Tüm kısayolları göster
* Q: Kapama uyarısını aç

## 6. Edit Mode Kısayolları

* Ctrl + Home: Başa git
* Ctrl + End: Sona git
* Ctrl + A: Tümü seç
* Ctrl + Left/Right: Kelime bazlı hareket
* Ctrl + Shift + P: Command Palette

# GOOGLE COLAB KLAVYE KISAYOLLARI

Colab, Jupyter’dan türemiştir fakat aynı değildir.
Kısayollar **çok daha fazladır**.

## 1. Çalıştırma

* Shift + Enter: Çalıştır ve aşağı geç
* Ctrl + Enter: Çalıştır ve kal
* Alt + Enter: Çalıştır ve yeni hücre ekle
* Ctrl + M + Enter: Tüm hücreleri çalıştır
* Ctrl + M + . : Çalışmayı durdur (interrupt)
* Ctrl + M + 0: Runtime restart

## 2. Hücre Yönetimi

* Ctrl + M + B: Alta yeni hücre
* Ctrl + M + A: Üste yeni hücre
* Ctrl + M + D: Hücre sil
* Ctrl + M + M: Markdown yap
* Ctrl + M + Y: Code yap
* Ctrl + M + H: Kod gizle/göster
* Ctrl + M + L: Satır numarası aç/kapa

## 3. Bölme ve Birleştirme

* Ctrl + M + - : Hücre böl
* Ctrl + M + J: Sonraki hücreye ince çizgi ile bağlan
* Ctrl + M + K: Önceki hücreye ince çizgi ile bağlan

## 4. Navigasyon

* Ctrl + M + Up / Down: Hücreler arası dolaş
* Ctrl + M + ] : Indent
* Ctrl + M + [ : Unindent

## 5. Düzenleme

* Ctrl + / : Yorum satırı aç/kapa
* Ctrl + Shift + K: Satır yukarı taşı
* Ctrl + Shift + J: Satır aşağı taşı
* Ctrl + ] : Indent
* Ctrl + [ : Unindent

## 6. Arama & Menü

* Ctrl + Shift + F: Belgede arama
* Ctrl + Shift + P: Komut paleti
* Ctrl + Alt + Shift + H: Tüm kısayollar

## 7. Genel

* Ctrl + M + S: Kaydet
* Ctrl + M + E: Çıkış (runtime durumu)
* Ctrl + M + T: En üste git
* Ctrl + M + B: Aşağı kaydır
