# Pandas ile Veri Manipülasyonu ve Analizi

Pandas, Python'da veri analizi ve manipülasyonu için vazgeçilmez bir kütüphanedir. Özellikle tablo biçimindeki verilerle (DataFrame'ler) çalışmak için tasarlanmıştır. Bu dokümanda, Pandas'ın en sık kullanılan ve güçlü özelliklerinden bazılarını detaylıca inceleyeceğiz.

## 1. `groupby()` Kullanımı

**`groupby()`** fonksiyonu, veri çerçevesini (DataFrame) belirli kriterlere göre **gruplara ayırmak** ve ardından her grup üzerinde bağımsız olarak işlemler yapmak için kullanılır. SQL'deki `GROUP BY` ifadesine benzer bir işlevi vardır.

**İşleyiş:**

1.  **Bölme (Splitting):** DataFrame, bir veya daha fazla sütunun benzersiz değerlerine göre parçalara (gruplara) ayrılır.
2.  **Uygulama (Applying):** Her bir bağımsız grup üzerinde bir fonksiyon (toplama, ortalama alma, sayma vb.) uygulanır.
3.  **Birleştirme (Combining):** Her grup için uygulanan fonksiyonun sonuçları tek bir sonuç veri yapısında birleştirilir.

**Örnek:**

```python
import pandas as pd
import numpy as np

data = {'Bölge': ['Kuzey', 'Güney', 'Kuzey', 'Doğu', 'Güney', 'Batı', 'Doğu', 'Kuzey'],
        'Ürün': ['A', 'B', 'A', 'C', 'B', 'A', 'C', 'B'],
        'Satış': [100, 150, 120, 80, 200, 110, 90, 130],
        'Maliyet': [50, 70, 60, 40, 90, 55, 45, 65]}
df = pd.DataFrame(data)

print("Orjinal DataFrame:")
print(df)
print("\n" + "="*50 + "\n")

# Tek bir sütuna göre gruplama ve toplam satışları bulma
toplam_satis_bolgeye_gore = df.groupby('Bölge')['Satış'].sum()
print("Bölgelere göre toplam satışlar:\n", toplam_satis_bolgeye_gore)
print("\n" + "="*50 + "\n")

# Birden fazla sütuna göre gruplama ve ortalama satışları bulma
ortalama_satis_bolge_urun_gore = df.groupby(['Bölge', 'Ürün'])['Satış'].mean()
print("Bölge ve Ürüne göre ortalama satışlar:\n", ortalama_satis_bolge_urun_gore)
print("\n" + "="*50 + "\n")
```

## 2. `agg()` Kullanımı

**`agg()`** (aggregate) fonksiyonu, DataFrame'inizdeki bir veya daha fazla sütun üzerinde **birden çok farklı toplama işlemi** yapmanızı sağlar. Genellikle `groupby()` ile birlikte kullanılarak gruplanmış veriler üzerinde özet istatistikler oluşturmak için tercih edilir.

**Kullanım Çeşitleri:**

* **Tek sütun, birden fazla işlem:** Bir sütun üzerinde aynı anda ortalama, standart sapma gibi birden fazla istatistik hesaplama.
* **Farklı sütunlarda farklı işlemler:** Her sütuna özel toplama fonksiyonları atama.
* **`groupby()` ile:** Gruplanmış veriler için çok yönlü özetler oluşturma.

**Örnek:**

```python
# 'Bölge'ye göre gruplayıp 'Satış' ve 'Maliyet' için farklı istatistikler alma
grup_istatistikleri = df.groupby('Bölge').agg(
    toplam_satis=('Satış', 'sum'),           # Satışların toplamı
    ortalama_maliyet=('Maliyet', 'mean'),   # Maliyetlerin ortalaması
    satis_sayisi=('Satış', 'count'),         # Satışların adedi
    en_yuksek_satis=('Satış', 'max')         # En yüksek satış
)

print("Bölgelere göre çeşitli istatistikler (agg kullanımı):\n", grup_istatistikleri)
print("\n" + "="*50 + "\n")

# Tek sütun üzerinde birden fazla işlem (groupby olmadan)
deger_istatistikleri = df['Satış'].agg(['sum', 'mean', 'std'])
print("Satış sütunu için toplam, ortalama ve standart sapma:\n", deger_istatistikleri)
print("\n" + "="*50 + "\n")
```

## 3. `merge()` Kullanımı

**`merge()`** fonksiyonu, SQL'deki `JOIN` işlemlerine benzer şekilde, **iki DataFrame'i bir veya daha fazla ortak sütuna (anahtar) göre birleştirmek** için kullanılır. Farklı birleştirme türlerini (inner, outer, left, right) destekler.

**Temel Parametreler:**

* **`right`:** Birleştirilecek ikinci DataFrame.
* **`how`:** Birleştirme türü (`'inner'`, `'outer'`, `'left'`, `'right'`). Varsayılan `'inner'`.
* **`on`:** Birleştirme için kullanılacak sütun veya sütunların adı (eğer her iki DataFrame'de de aynı isimdeyse).
* **`left_on` / `right_on`:** Eğer birleştirme sütunları farklı isimdeyse, sol ve sağ DataFrame'deki sütun adları.

**Örnek:**

```python
# İkinci bir DataFrame oluşturalım
urun_bilgileri = pd.DataFrame({
    'Ürün': ['A', 'B', 'C', 'D'],
    'Kategori': ['Elektronik', 'Giyim', 'Gıda', 'Elektronik'],
    'Fiyat': [1000, 200, 50, 500]
})

print("Ürün Bilgileri DataFrame:")
print(urun_bilgileri)
print("\n" + "="*50 + "\n")

# 'df' ile 'urun_bilgileri' DataFrame'lerini 'Ürün' sütununa göre birleştirme (inner join)
birlesik_df = pd.merge(df, urun_bilgileri, on='Ürün', how='inner')
print("df ve urun_bilgileri DataFrame'lerinin 'Ürün'e göre birleşimi (inner):\n", birlesik_df)
print("\n" + "="*50 + "\n")

# Sol birleştirme (left join) - df'deki tüm kayıtlar ve eşleşen ürün bilgileri
sol_birlesim = pd.merge(df, urun_bilgileri, on='Ürün', how='left')
print("df ve urun_bilgileri DataFrame'lerinin 'Ürün'e göre sol birleşimi:\n", sol_birlesim)
print("\n" + "="*50 + "\n")
```

## 4. `transform()` Kullanımı

**`transform()`** fonksiyonu, `groupby()` ile birlikte kullanıldığında çok faydalıdır. Bir gruba ait değerler üzerinde bir hesaplama yapar ve bu **sonucu orijinal DataFrame'in boyutuyla aynı olacak şekilde geri döndürür**. Bu, gruplama sonrası sonuçları orijinal DataFrame'e yeni sütunlar olarak eklemek için idealdir.

**`agg()` ile Farkı:**

* `agg()`: Her grup için **tek bir özet değer** döndürür, bu da orijinal DataFrame'den daha küçük bir çıktıya neden olur.
* `transform()`: Her grubun hesaplanmış değerini **orijinal DataFrame'in her bir satırına karşılık gelecek şekilde** döndürür, böylece orijinal DataFrame ile aynı boyutta bir Series veya DataFrame elde edilir.

**Örnek:**

```python
# Her bölgedeki satışların ortalamasını hesapla ve orijinal DataFrame'e ekle
df['Bölge_Ort_Satış'] = df.groupby('Bölge')['Satış'].transform('mean')

print("Bölgelere göre ortalama satışlar, transform ile orijinal DataFrame'e eklendi:\n", df)
print("\n" + "="*50 + "\n")

# Her bölgedeki satışların toplamını hesapla ve orijinal DataFrame'e ekle
df['Bölge_Toplam_Satış'] = df.groupby('Bölge')['Satış'].transform('sum')

print("Bölgelere göre toplam satışlar, transform ile orijinal DataFrame'e eklendi:\n", df)
print("\n" + "="*50 + "\n")
```

## 5. `concat()` Kullanımı

**`concat()`** fonksiyonu, Pandas nesnelerini (Series veya DataFrame'ler) **dikey veya yatay olarak birleştirmek** için kullanılır. `merge()`'den farklı olarak, `concat()` anahtar sütunlara ihtiyaç duymaz; daha çok yığınlama veya yan yana birleştirme işlemi yapar.

**Temel Parametreler:**

* **`objs`:** Birleştirilecek Pandas nesnelerinin listesi.
* **`axis`:** Birleştirme ekseni. `0` (varsayılan) dikey birleştirme (satır ekleme), `1` ise yatay birleştirme (sütun ekleme).
* **`join`:** Birleştirme türü (`'outer'` varsayılan, `'inner'`).
* **`ignore_index`:** `True` ise, sonuç DataFrame'in indeksini sıfırlar.

**Örnek:**

```python
# İki yeni DataFrame oluşturalım
df1 = pd.DataFrame({'A': [1, 2], 'B': [3, 4]})
df2 = pd.DataFrame({'A': [5, 6], 'B': [7, 8]})
df3 = pd.DataFrame({'C': [9, 10], 'D': [11, 12]})

print("df1:\n", df1)
print("\ndf2:\n", df2)
print("\ndf3:\n", df3)
print("\n" + "="*50 + "\n")

# Dikey birleştirme (satır ekleme)
dikey_birlesim = pd.concat([df1, df2])
print("df1 ve df2'nin dikey birleşimi:\n", dikey_birlesim)
print("\n" + "="*50 + "\n")

# Dikey birleştirme ve indeksi sıfırlama
dikey_birlesim_sifirlanmis_index = pd.concat([df1, df2], ignore_index=True)
print("df1 ve df2'nin dikey birleşimi (indeks sıfırlanmış):\n", dikey_birlesim_sifirlanmis_index)
print("\n" + "="*50 + "\n")

# Yatay birleştirme (sütun ekleme)
yatay_birlesim = pd.concat([df1, df3], axis=1)
print("df1 ve df3'ün yatay birleşimi:\n", yatay_birlesim)
print("\n" + "="*50 + "\n")
```

## 6. Pivot Tablosu Kullanımı (`pivot_table()`)

Pandas'taki **`pivot_table()`** fonksiyonu, Excel'deki pivot tablolarına benzer şekilde çalışır. Verileri özetlemek, yeniden şekillendirmek ve farklı perspektiflerden analiz etmek için kullanılır. Genellikle büyük ve karmaşık veri setlerinden özetlenmiş, okunabilir tablolar oluşturmak için idealdir.

**Temel Parametreler:**

* **`values`:** Pivot tablosunda gösterilecek değerleri içeren sütun(lar).
* **`index`:** Yeni tablonun satır indeksini oluşturacak sütun(lar).
* **`columns`:** Yeni tablonun sütun başlıklarını oluşturacak sütun(lar).
* **`aggfunc`:** Uygulanacak toplama fonksiyonu (örneğin, `'mean'`, `'sum'`, `np.sum`, `len`). Varsayılan `'mean'`.
* **`fill_value`:** Eksik değerleri (NaN) doldurmak için kullanılacak değer.

**Örnek:**

```python
print("Orjinal DataFrame (tekrar gösterim):\n", df)
print("\n" + "="*50 + "\n")

# Bölgelere göre ürünlerin ortalama satışlarını gösteren bir pivot tablosu
pivot_satis_ortalama = pd.pivot_table(df, values='Satış', index='Bölge', columns='Ürün', aggfunc='mean')
print("Bölgelere göre ürünlerin ortalama satışları (Pivot Tablo):\n", pivot_satis_ortalama)
print("\n" + "="*50 + "\n")

# Bölgelere göre ürünlerin toplam satışlarını gösteren ve NaN değerleri 0 ile dolduran pivot tablosu
pivot_satis_toplam = pd.pivot_table(df, values='Satış', index='Bölge', columns='Ürün', aggfunc='sum', fill_value=0)
print("Bölgelere göre ürünlerin toplam satışları (Pivot Tablo, NaN=0):\n", pivot_satis_toplam)
print("\n" + "="*50 + "\n")
```

## 7. Tablonun CSV Olarak Kaydedilmesi (`to_csv()`)

Veri analizi sonuçlarınızı veya manipüle ettiğiniz DataFrame'i daha sonra kullanmak veya başkalarıyla paylaşmak için bir CSV (Comma Separated Values) dosyasına kaydetmek yaygın bir işlemdir. Pandas'taki **`to_csv()`** fonksiyonu bu görevi kolayca yerine getirir.

**Temel Parametreler:**

* **`path_or_buf`:** Dosyanın kaydedileceği yol ve adı (örn: `'analiz_sonuclari.csv'`).
* **`sep`:** Ayırıcı karakter (varsayılan virgül `,`).
* **`index`:** `True` ise DataFrame indeksi de dosyaya kaydedilir. `False` ise kaydedilmez.
* **`header`:** `True` ise sütun başlıkları kaydedilir.
* **`encoding`:** Dosya kodlaması (örn: `'utf-8'`).

**Örnek:**

```python
# Önceki pivot tablo sonucunu bir CSV dosyasına kaydedelim
output_filename = 'bolge_urun_satis_ozeti.csv'
pivot_satis_toplam.to_csv(output_filename, index=True, encoding='utf-8')

print(f"'{output_filename}' dosyası başarıyla kaydedildi.")
```
