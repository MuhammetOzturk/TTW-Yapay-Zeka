### 3. Asal Sayı Kontrol Fonksiyonu:
Verilen bir sayının asal olup olmadığını kontrol eden `asal_mı(sayı)` adında bir fonksiyon yazın. Asal sayılar sadece 1'e ve kendisine bölünebilen sayılardır. Fonksiyon `True` veya `False` döndürmelidir.

```python
asal_mı(7) -> True
asal_mı(10) -> False
 ```

````python
def asal_mi(sayi):

    asalmi = True
    if sayi < 2:
        return False

    for i in range(2,sayi):
        if sayi % i == 0:
            asalmi = False
            break

    return asalmi

sayi = int(input("Bir sayi giriniz : "))
sonuc = asal_mi(sayi)
if sonuc:
    print("Asaldir")
else:
    print("Asal degildir")
````

### 1. Faktoriyel Hesaplama Fonksiyonu:
Verilen bir pozitif tam sayının faktoriyelini hesaplayan `faktoriyel_hesapla(sayi)` adında bir fonksiyon yazın. Faktoriyel, 1'den o sayıya kadar olan tüm tam sayıların çarpımıdır (örneğin, 5! = 5 * 4 * 3 * 2 * 1 = 120).

*Örnek: `faktoriyel_hesapla(5)` -> `120`*

````python
def faktoriyel_hesapla(sayi):
    faktoriyel = 0
    for i in range(1,sayi+1):
        faktoriyel *= i
    return faktoriyel
````

### 2. Tek ve Çift Sayıları Ayırma Fonksiyonu:
Bir tamsayı listesi alan ve bu listedeki tek sayıları ayrı bir liste olarak, çift sayıları ayrı bir liste olarak döndüren `tek_cift_ayir(liste)` adında bir fonksiyon oluşturun.

*Örnek: `tek_cift_ayir([1, 2, 3, 4, 5, 6])` -> `([1, 3, 5], [2, 4, 6])`*

````python
def tek_cift_ayir(liste):
    tek = []
    cift = []
    for i in liste:
        if i%2==0:
            cift.append(i)
        else:
            tek.append(i)
    return tek,cift

tek_cift_ayir([3,5,8,10,1,9])
````

### 3. Dizelerdeki Sesli Harfleri Sayma Fonksiyonu:
Verilen bir metin içerisindeki (büyük/küçük harf duyarsız) sesli harflerin ('a', 'e', 'ı', 'i', 'o', 'ö', 'u', 'ü') toplam sayısını döndüren `sesli_harf_sayisi(metin)` adında bir fonksiyon yazın.

````python
unlu_harfler = ['a', 'e', 'ı', 'i', 'o', 'ö', 'u', 'ü']
metin = "MerhabA Dunya"
metin = metin.lower()
sayac = 0
for harf in metin:
    if harf in unlu_harfler:
        sayac += 1
sayac
````

### 4. Maksimum ve Minimum Değer Bulma Fonksiyonu:
Bir sayı listesi alan ve listedeki en büyük (maksimum) ve en küçük (minimum) değerleri bir tuple olarak döndüren `max_min_bul(liste)` adında bir fonksiyon oluşturun. Boş bir liste verilirse uygun bir hata mesajı döndürmelidir.

*Örnek: `max_min_bul([10, 5, 20, 15, 3])` -> `(20, 3)`*

````python
def max_min_bul(liste):
    if not liste:
        return "Bos liste"
    maxVal = max(liste)
    minVal = min(liste)
    return maxVal,minVal

max_min_bul([10, 5, 20, 15, 3])
````

### 5. Fibonacci Serisi Fonksiyonu:
Verilen bir `n` pozitif tam sayısı için, Fibonacci serisinin ilk `n` terimini içeren bir liste döndüren `fibonacci_serisi(n)` adında bir fonksiyon yazın. Fibonacci serisi 0 ve 1 ile başlar, sonraki her terim önceki iki terimin toplamıdır (0, 1, 1, 2, 3, 5, ...).

*Örnek: `fibonacci_serisi(6)` -> `[0, 1, 1, 2, 3, 5]`*

````python
def fibonacci_serisi(n):
    fibo = [0,1]
    if n < 2:
        return fibo

    for _ in range(n-2):
        sonraki_deger = fibo[-1] + fibo[-2]
        fibo.append(sonraki_deger)

    return fibo

fibonacci_serisi(8)
fibonacci_serisi(2)
fibonacci_serisi(1)
````

### 5. Ortalama Hesaplama Fonksiyonu:
Sayılar listesi alan ve bu sayıların ortalamasını döndüren `ortalama_hesapla(sayılar)` adında bir fonksiyon yazın. Liste boşsa uygun bir mesaj döndürmelidir.

*Örnek: ortalama_hesapla([10, 20, 30, 40]) -> 25.0*

````python
def ortalama_hesapla(sayilar):
    if ''.join(sayilar) == '':
        return "Bos liste"

    toplam = 0
    for i in sayilar:

        sayi = int(i)
        toplam += sayi
        print(f"Toplam : {toplam}")

    return toplam / len(sayilar)

ortalama_hesapla(["2","4","6"])
````
