```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)
plt.plot(x, y)
plt.show()
```

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y, '*')
plt.grid(True)
plt.xlabel('x ekseni')
plt.ylabel('sin(x) ekseni')
plt.title('Sinüs Fonksiyonu')
plt.show()
```

```python
plt.plot(x, y, color='red', linestyle = '-.')
plt.grid(True)
plt.xlabel('x ekseni')
plt.ylabel('sin(x) ekseni')
plt.title('Sinüs Fonksiyonu')
plt.show()
```

```python
y2 = np.cos(x)
plt.plot(x, y, label = 'sin(x)',markersize=10,marker='.')
plt.plot(x, y2, label = 'cos(x)')

plt.grid(True)
plt.xlabel('x ekseni')
plt.ylabel('sin(x) ekseni')
plt.title('Sinüs Fonksiyonu')

#Hangi cizginin neye karsilik geldigini belirtir
plt.legend()
plt.show()
```

```python
#Dagilim grafigi
np.random.seed(None)

color = np.zeros(10)
color = np.random.randint(0,255,10)

np.random.seed(10)

x = np.random.rand(10)
y = np.random.rand(10)

#c ==> colormap

#c renklendirme paleti
#s noktalarin boyutu

#Her bir nokta icin tek renk
#plt.scatter(x,y,c='red' , s=90)

#Her bir nokta icin farkli renkler
plt.scatter(x,y,c=color , s=np.arange(100,500,step=40))
plt.figure(figsize=(1,10))
plt.show()
```

```python
fig, ax = plt.subplots(1,2)

x = np.linspace(0,10,100)
y = np.sin(x)
y2 = np.cos(x)

ax[0].plot(x,y)
ax[1].plot(x,y2)

ax[0].set_title('Sinüs Fonksiyonu')
ax[1].set_title('Cosinüs Fonksiyonu')

ax[0].set_xlabel('x ekseni')
ax[0].set_ylabel('sin(x) ekseni')

ax[1].set_xlabel('x ekseni')
ax[1].set_ylabel('cos(x) ekseni')

fig.set_figwidth(8)
fig.set_figheight(3)

#Bu fonksiyon cizim alanlari arasinda uygun bosluklari olusturur
plt.tight_layout()
plt.show()
```

```python
from PIL import Image
import requests
from io import BytesIO

url = 'https://www.elityavru.com/images/blog/munchkin-kedisi.webp'
response = requests.get(url)

img = Image.open(BytesIO(response.content))
img = img.resize((300,200))
img
```

```python

fig, axs = plt.subplots(2,2)
plt.suptitle('Grafikler')

x = np.linspace(0,10,100)
y = np.sin(x)
y2 = np.cos(x)

axs[0,0].plot(x,y, label='sin(x)')
axs[0,1].plot(x,y2, label='cos(x)')
axs[1,0].imshow(img)

axs[0,0].legend()

axs[0,1].set_ylim(-1,1.5)
axs[0,1].legend(loc=[0.11,0.7])

axs[0,0].set_title('Sinüs Fonksiyonu')
axs[0,1].set_title('Cosinüs Fonksiyonu')
axs[1,0].set_title('Kedi')



axs[1,0].xaxis.set_visible(False)
axs[1,0].yaxis.set_visible(False)

axs[1,1].annotate(xy=[0.1,0.7],text='Merhaba Dunya')

plt.tight_layout()

plt.show()

plt.style.use('seaborn-v0_8-notebook')
fig.savefig('grafikler.png')
```

```python

veri = np.random.normal(loc=0, scale=1, size=100)

x = np.arange(1,len(veri)+1)

fig , axs = plt.subplots(2,2,figsize=(10,6))

fig.suptitle('Grafikler')
fig.set_figwidth(15)
fig.set_figheight(8)

axs[0,0].plot(x,veri)
axs[0,0].set_title('Orjinal Veri')

axs[0,1].hist(veri, bins=20,edgecolor = 'black')
axs[0,1].set_title('Histogram')

axs[1,0].boxplot(veri)
axs[1,0].set_title('Boxplot')

axs[1,1].scatter(x,veri)
axs[1,1].set_title('Scatter Plot')


plt.grid(True)
plt.tight_layout()

plt.show()
```

```python
plt.style.use('bmh')

veri = np.random.normal(loc=0, scale=1, size=10)

x = np.arange(1,len(veri)+1)

fig, axs = plt.subplot_mosaic("""
AB
CC
""")

fig.suptitle('Grafikler')

axs['A'].plot(x,veri)
axs['A'].set_title('Orjinal Veri')

axs['B'].hist(veri, bins=10)
axs['B'].set_title('Histogram')

axs['C'].boxplot(veri, vert=False)
axs['C'].set_title('Boxplot')


plt.grid(True)
plt.tight_layout()

plt.show()
```

```python
fig, axs = plt.subplot_mosaic("""
AD
BD
CD
""", figsize=(10,8))

axs['A'].plot(x,veri)
axs['A'].set_ylim(-3,3)
axs['A'].set_title('Orjinal Veri')

axs['B'].set_ylim(-3,4)
axs['B'].hist(veri, bins=10)

axs['B'].set_title('Histogram')

axs['C'].set_ylim(-1,1)
axs['C'].scatter(x,veri)

axs['D'].boxplot(veri, vert=True)
axs['D'].set_title('Boxplot')
axs['D'].set_xlim(0,2)



fig.suptitle('Grafikler')

plt.grid(True)
plt.tight_layout()

plt.show()
```
