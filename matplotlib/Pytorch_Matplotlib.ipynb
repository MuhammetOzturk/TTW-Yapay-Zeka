# Bar Grafigi

````python
kategoriler = ['A','B','C','D','E']
degerler    = [50,40,20,10,30]
plt.bar(kategoriler,degerler,color = "#00FFFF", edgecolor = "black")
plt.title("Baslik")
plt.xlabel("x ekseni")
plt.ylabel("y ekseni")
plt.show()
````

````python
kategoriler = ['A','B','C','D','E']
degerler    = [50,40,20,10,30]
plt.bar(kategoriler,degerler,color = "#00FFFF", edgecolor = "black")
plt.title("Baslik")
plt.xlabel("x ekseni")
plt.ylabel("y ekseni")
plt.grid(True)
plt.show()
````

````python
fig, (ax1,ax2) = plt.subplots(1,2)

fig.set_figwidth(10)
fig.set_figheight(5)

ax1.bar(kategoriler,degerler,color = "#00FF90", edgecolor = "black")
ax1.set_title("Baslik")
ax1.set_xlabel("x ekseni")
ax1.set_ylabel("y ekseni")
ax1.grid(axis='x')

ax2.bar(kategoriler,degerler,color = "#00FFFF", edgecolor = "black")
ax2.set_title("Baslik")
ax2.set_xlabel("x ekseni")
ax2.set_ylabel("y ekseni")
ax2.grid(axis='y')

plt.suptitle("Bar Grafigi")
plt.show()
````

````python
fig , (ax1,ax2) = plt.subplots(1,2)

fig.set_figwidth(10)
fig.set_figheight(5)

ax1.bar(kategoriler,degerler,color = "#00FFFF", edgecolor = "black")
ax2.barh(kategoriler,degerler,color = "#00FF90", edgecolor = "black")

plt.show()
````

# Pasta Grafigi

````python
print(f"Kategoriler : {kategoriler}")
print(f"Degerler   : {degerler}")
````

````python
plt.pie(degerler,labels=kategoriler,autopct="%1.1f%%",shadow=False,)
plt.title("Pasta Grafigi")
plt.axis("equal")
plt.show()
````

````python
plt.pie(degerler,labels=kategoriler,colors=['orange','skyblue','yellow','red','green'],
        wedgeprops={"edgecolor":"black","linewidth":1},autopct="%1.1f%%",shadow=False,)

plt.title("Pasta Grafigi")
plt.axis("equal")
plt.show()
````

# Histogram Grafigi

````python
import random

random.seed(1)

liste = list("ABCDEF")
print(f"Liste : {liste}")

rnd_liste = random.choices(liste,k=20)
print(f"Random Liste : {rnd_liste}")

plt.hist(rnd_liste,bins=5 ,edgecolor="black", color="#00FAFF")
plt.grid(axis="y")
plt.title("Histogram Grafigi")
plt.show()
````

````python
fig, ax = plt.subplots(1,1)

x = np.linspace(0,10,100)
sin = np.sin(x)
cos = np.cos(x)

ax.plot(x,sin,label="sin(x)")
ax.plot(x,cos,label="cos(x)")

ax.legend()
ax.set_title("Sinüs Fonksiyonu")
ax.set_xlabel("x")
ax.set_ylabel("sin(x)")
````

````python
import cv2
import requests

url = 'https://www.startpage.com/av/proxy-image?piurl=https%3A%2F%2Fwww.encyclopedia-titanica.org%2Fimages%2Felisabeth_walton_allen_1912_H.jpg&sp=1751355641T35cf493401d205e069cbc4fffab8f5ea465a89ddae2ca49dc8da6bb732d5dab1'

response = requests.get(url)
if not response.ok:
    print("Istek basarisiz!")
else:
    with open("image.jpg","wb") as f:
        f.write(response.content)

    image = cv2.imread("image.jpg")
    image = cv2.cvtColor(image,cv2.COLOR_BGR2RGB)
    plt.imshow(image)
    plt.axis("off")
    plt.title("Elisabeth Walton Allen")
    plt.show()
````

````python
fig, (ax1,ax2) = plt.subplots(1,2)

fig.set_figwidth(10)
fig.set_figheight(4)

ax1.bar(["Yas","Ortalama Gelir"],[30,100])
ax1.annotate("Yas",xy=(0,33))
ax1.annotate("Ortalama Gelir",xy=(1,103))
ax1.set_ylim(0,110)
ax1.set_xlim(-1,2)


ax2.imshow(image)
ax2.axis("off")

plt.show()
````
