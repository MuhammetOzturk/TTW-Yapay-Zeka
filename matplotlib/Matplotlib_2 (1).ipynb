**`scatter`** fonksiyonu parametreleri.  

| Parametre      | Açıklama                                                                               |
| -------------- | -------------------------------------------------------------------------------------- |
| `x`            | Noktaların x-koordinatları (liste, NumPy dizisi, pandas serisi).                       |
| `y`            | Noktaların y-koordinatları (liste, NumPy dizisi, pandas serisi).                       |
| `s`            | Noktaların boyutları. Tek bir sayı (hepsi aynı boyda) veya dizi olabilir.              |
| `c`            | Renk veya renk dizisi. Tek bir renk adı, RGB kodu ya da bir renk dizisi olabilir.      |
| `marker`       | Nokta şekli. (Örn: `'o'`, `'^'`, `'s'`, `'x'` vs.)                                     |
| `alpha`        | Saydamlık derecesi (0 ile 1 arasında). 0 tamamen şeffaf, 1 tamamen opak.               |
| `cmap`         | Renk haritası. `c` parametresi ile birlikte kullanılır. (Örn: `'viridis'`, `'plasma'`) |
| `edgecolors`   | Noktaların kenar renkleri. (Örn: `'black'`, `'none'`)                                  |
| `linewidths`   | Kenar çizgisi kalınlığı.                                                               |
| `label`        | Legend (etiket) için ad. `plt.legend()` ile görünür yapılabilir.                       |
| `norm`         | Renk skalasını normalize etmek için `Normalize` nesnesi.                               |
| `vmin`, `vmax` | Renk değerleri için minimum ve maksimum değerler. `c` parametresiyle kullanılır.       |
| `data`         | `x` ve `y` isimlerinin anahtar olduğu bir veri sözlüğü veya DataFrame.                 |


````python
np.random.seed(1)
data = {'a': np.arange(50),
        'c': np.random.randint(0, 50, 50),
        'd': np.random.randn(50)}
data['b'] = data['a'] + 10 * np.random.randn(50)
data['d'] = np.abs(data['d']) * 100

fig, ax = plt.subplots(figsize=(5, 2.7), layout='constrained')

ax.scatter('a', 'b', c='c', s='d', data=data)
ax.set_xlabel('Zaman')
ax.set_ylabel('Deger')

plt.show()
````

````python
def my_plotter(ax, data1, data2, param_dict):
    """
    Yardımcı fonksiyon.
    """
    out = ax.plot(data1, data2, **param_dict)
    return out

#data1 = np.linspace(0,180,60)
data1 = np.arange(0,270,step=3)
data2 = np.sin(data1)
fig, ax = plt.subplots(figsize=(5, 2.7), layout='constrained')

my_plotter(ax, data1, data2, {'marker': 'o',})

plt.show()
````

````python
data1, data2, data3, data4 = np.random.randn(4, 100)  # 4 tane veri seti olustur
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(5, 2.7))
my_plotter(ax1, data1, data2, {'marker': 'x'})
my_plotter(ax2, data3, data4, {'marker': 'o'})
plt.show()
````

# Annotations - Aciklamalar

````python
x = np.arange(0,5,0.3)
y = np.sin(x)

fig, ax = plt.subplots(figsize=(5, 2.7),tight_layout=True)
ax.plot(x,y)

plt.show()
#Soru en tepe noktaya aciklama nasil eklenir.
````

````python
x = np.arange(0,20,0.2)
y = np.sin(x)

uniq = np.unique(y)

uniq.sort()
ilk_uc_buyuk_nokta = uniq[-3:][::-1]

for i,mx in enumerate(ilk_uc_buyuk_nokta):
    indisler = np.where(y==mx)
    for indis in indisler:
        print(x[indis],y[indis])
        plt.annotate(f"{i+1}.nokta",(x[indis],y[indis]))

plt.plot(x,y)
plt.tight_layout()
plt.show()
````

````python
x = np.arange(0,20,0.2)
y = np.sin(x)

uniq = np.unique(y)

uniq.sort()
ilk_uc_buyuk_nokta = uniq[-3:][::-1]

plt.ylim(-1.5,1.5)
for i,mx in enumerate(ilk_uc_buyuk_nokta):
    indisler = np.where(y==mx)
    for indis in indisler:
        plt.annotate(f"{i+1}.nokta",
            xy=(x[indis],y[indis]),
            xytext=(x[indis],y[indis]+0.4),
            arrowprops=dict(facecolor='red',shrink=0.05)
        )

plt.plot(x,y,linestyle='--')

plt.tight_layout()
plt.show()


#Soru 2.nokta ile 3.nokta arasina duz cizgi nasil cekilir
````

````python
x = np.arange(0,20,0.2)
y = np.sin(x)

uniq = np.unique(y)

uniq.sort()
ilk_uc_buyuk_nokta = uniq[-3:][::-1]

plt.ylim(-1.5,1.5)
for i,mx in enumerate(ilk_uc_buyuk_nokta):
    indisler = np.where(y==mx)
    for indis in indisler:
        plt.annotate(f"{i+1}.nokta",
            xy=(x[indis],y[indis]),
            xytext=(x[indis],y[indis]+0.4),
            arrowprops=dict(facecolor='red',shrink=0.05)
        )

plt.plot(x,y,linestyle='--')

noktalar = ilk_uc_buyuk_nokta[1:]
print(f"Noktalar : {noktalar}")

a = []
b = []

for nokta in noktalar:
    indisler = np.where(y==nokta)
    for indis in indisler:

        a.append(x[indis])
        b.append(y[indis])

plt.plot(a,b)

plt.plot(x,y,linestyle='--')
plt.tight_layout()
plt.show()
````

Elbette! `tick_params()` fonksiyonu, **matplotlib** kütüphanesinde eksen üzerindeki **tick (çentik)** işaretlerinin ve etiketlerinin görünümünü özelleştirmek için kullanılır.

---

## ✅ Fonksiyonun Genel Yapısı

```python
Axes.tick_params(axis='both', **kwargs)
```

---

## 🔹 Önemli Parametreler

| Parametre                                            | Açıklama                                                           |
| ---------------------------------------------------- | ------------------------------------------------------------------ |
| `axis`                                               | `'x'`, `'y'`, veya `'both'` — hangi eksene uygulanacağı            |
| `which`                                              | `'major'`, `'minor'`, `'both'` — hangi tür çentiklere uygulanacağı |
| `direction`                                          | `'in'`, `'out'`, `'inout'` — çizgi yönü                            |
| `length`                                             | Çizgi uzunluğu                                                     |
| `width`                                              | Çizgi kalınlığı                                                    |
| `color`                                              | Çizgi rengi                                                        |
| `labelsize`                                          | Etiket yazı boyutu                                                 |
| `labelcolor`                                         | Etiket yazı rengi                                                  |
| `bottom`, `top`, `left`, `right`                     | `True`/`False` — o kenarda çentik olup olmayacağı                  |
| `labelbottom`, `labeltop`, `labelleft`, `labelright` | Etiketlerin gösterilip gösterilmeyeceği                            |

---

## 🔸 Örnekler

### 1. X ekseni çentiklerini kırmızı yap, etiketleri büyüt:

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()
ax.plot([1, 2, 3], [1, 4, 9])
ax.tick_params(axis='x', direction='inout', length=10, width=2, color='red', labelsize=14)
plt.show()
```

### 2. Sadece Y ekseni etiketlerini gizle:

```python
ax.tick_params(axis='y', labelleft=False)
```

### 3. Hem x hem y ekseninde küçük çentikleri özelleştir:

```python
ax.tick_params(axis='both', which='minor', length=4, color='gray')
```

---

## 🎯 İpucu

Küçük (minor) çentiklerin görünmesi için önce aşağıdaki gibi etkinleştirmek gerekebilir:

```python
ax.minorticks_on()
```

# Arranging multiple Axes in a Figure

Cizimler Axes nesneleri kullanilarak gerceklestirilir.  
Axes nesneleri ise Axis nesnelerinden olusur.

````python
#subplots kullanimi

#Cizim alani olusturuluyor.
fig = plt.figure(figsize=(8,6), tight_layout=True)

fig.suptitle('Grafik Başlığı')

#Cizim alaninda eksenler yerlestiriliyor.
axs = fig.subplots(2, 2)
for i,ax in enumerate(axs.flat):
    ax.set_title(f'Alt Grafik {i+1}')
    ax.set_xlabel(f'x{i+1}')
    ax.set_ylabel(f'y{i+1}')
    ax.text(0.5,0.5,f'({i+1},{i+1})',ha='center',va='center')

ax.tick_params(axis='x',color='red',labelcolor='blue',
               width=3,length=10,labelsize=13)

plt.show()
````

````python
fig, axs = plt.subplots(2,2)

#axs.flat iteratif bir axes nesnesi doner
for i,ax in enumerate(axs.flat):
    if i % 2 == 0:
        ax.grid(True)
    else:
        ax.axhline(c='red',linewidth=5,linestyle=':')
    ax.annotate(f"ax({i})",(0.5,0.5),xycoords='axes fraction',ha='center',)

plt.show()
````

**fig.add_axes([sol, alt, genişlik, yükseklik])** metodu, şekle belirtilen [sol, alt, genişlik, yükseklik] dörtgeninde, normalleştirilmiş şekil koordinatlarında bir Axes (eksen) ekler. Her bir parametre şunları temsil eder:

sol: Axes'in sol kenarının şeklin sol kenarına olan uzaklığı.  
alt: Axes'in alt kenarının şeklin alt kenarına olan uzaklığı.  
genişlik: Axes'in genişliği.  
yükseklik: Axes'in yüksekliği.    
Tüm değerler, şeklin genişliğinin ve yüksekliğinin 0 ile 1 arasında değişen kesirleri olarak verilir.

````python
import matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec, GridSpecFromSubplotSpec

fig = plt.figure(figsize=(8, 4))

# Dış grid: 1 satır, 3 sütun (soldaki daha geniş)
outer_gs = GridSpec(1, 3, width_ratios=[1.5, 1, 1.5], figure=fig)

# Sol alanı 3 satıra böl, her biri farklı oranda yükseklikte
# Örneğin: üst = 3, orta = 1, alt = 2 oranında yükseklik
left_gs = GridSpecFromSubplotSpec(
    3, 1,                      # 3 satır, 1 sütun
    subplot_spec=outer_gs[0], # sadece sol alan
    height_ratios=[3, 1, 2],  # farklı yükseklikler
    hspace=0.7                # dikey boşluk
)

# Her parçayı subplot olarak ekle
axs_left = [fig.add_subplot(left_gs[i]) for i in range(3)]
axs_mid = fig.add_subplot(outer_gs[1])
axs_right = fig.add_subplot(outer_gs[2])

# Örnek başlıklar
axs_left[0].set_title("Sol üst (büyük)")
axs_left[1].set_title("Sol orta (küçük)")
axs_left[2].set_title("Sol alt (orta)")
axs_mid.set_title("Orta")
axs_right.set_title("Sağ")

plt.tight_layout()
plt.show()
````
