```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y)
plt.title("Sinüs Eğrisi")
plt.xlabel("x değeri")
plt.ylabel("sin(x)")
plt.grid(True)
plt.show()
```

```python
plt.plot([1, 2, 3], [4, 1, 7], linestyle='--', color='red', marker='o')
plt.show()
```

```python
kategoriler = ['A', 'B', 'C']
degerler = [10, 15, 7]

plt.bar(kategoriler, degerler, color='orange')
plt.show()
```

```python
plt.barh(kategoriler, degerler, color='skyblue', edgecolor='black')
```

```python
plt.pie(degerler, labels=kategoriler, autopct='%1.1f%%', startangle=90)
plt.axis('equal')  # Daire olarak göster
```

```python
veri = np.random.normal(loc=0, scale=1, size=1000)
plt.hist(veri, bins=10, color='purple', edgecolor='black')
```

```python
x = np.random.rand(50)
y = np.random.rand(50)
plt.scatter(x, y, c='green', marker='x')
```

```python
fig, axs = plt.subplots(2, 2, figsize=(10, 6))

axs[0, 0].plot(x, y)
axs[0, 1].bar(kategoriler, degerler)
axs[1, 0].hist(veri, bins=20)
axs[1, 1].scatter(x, y)

fig.suptitle("Alt Grafik Örnekleri")
plt.tight_layout()
```

```python
plt.title("Başlık")
plt.xlabel("X Eksen Başlığı")
plt.ylabel("Y Eksen Başlığı")
plt.legend(["Veri 1"])
plt.grid(True)
```
