```python
sayi = int(input("Bir sayi giriniz: "))

if sayi < 5:
    print(f"{sayi} 5'ten kucuktur")
elif sayi == 5:
    print(f"{sayi} 5'e esittir")
elif sayi > 10:
    print(f"{sayi} 10'dan buyuktur")
else:
    print(f"{sayi} 5'ten buyuktur")
```

### Problem:
Kullanıcıdan bir tam sayı almanızı ve bu sayının pozitif, negatif veya sıfır olup olmadığını belirten bir mesaj yazdırmanızı isteyen bir Python programı yazınız.

**Örnek Çıktılar:**
*   `Girdi: 5` -> `5 pozitif bir sayidir.`
*   `Girdi: -3` -> `-3 negatif bir sayidir.`
*   `Girdi: 0` -> `0 ne pozitif ne de negatiftir, sifirdir.`

```python
sayi = 10
if sayi > 10:
    print(f"{sayi} pozitif bir sayidir.")
elif sayi > 5:
    print(f"{sayi} 5 ile 10 arisinda bir sayidir.")
else:
    print(f"{sayi} 10'dan kucuk bir sayidir.")
```

### Eşitlik Durumları Örnekleri

```python
# == (Eşittir) Operatörü
a = 10
b = 10
c = 20

print(f"a == b: {a == b}") # True döndürür
print(f"a == c: {a == c}") # False döndürür
```

```python
# != (Eşit Değildir) Operatörü
x = 5
y = 7
z = 5

print(f"x != y: {x != y}") # True döndürür
print(f"x != z: {x != z}") # False döndürür
```

```python
x == y
x != y
x < y
x <= y
x > y
x >= y
```

```python
x = 1
y = 2
z = -1
0 < x <= y <= z
```

```python
# >= (Büyük veya Eşittir) Operatörü
num1 = 15
num2 = 10
num3 = 15

print(f"num1 >= num2: {num1 >= num2}") # True döndürür
print(f"num1 >= num3: {num1 >= num3}") # True döndürür
print(f"num2 >= num1: {num2 >= num1}") # False döndürür
```

```python
# <= (Küçük veya Eşittir) Operatörü
val1 = 8
val2 = 12
val3 = 8

print(f"val1 <= val2: {val1 <= val2}") # True döndürür
print(f"val1 <= val3: {val1 <= val3}") # True döndürür
print(f"val2 <= val1: {val2 <= val1}") # False döndürür
```
