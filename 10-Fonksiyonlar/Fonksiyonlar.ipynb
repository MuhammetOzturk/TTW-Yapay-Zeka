````python
def topla(*args):
    print(f"args : {args} tip: {type(args)}")
    toplam = 0
    for i in args:
        toplam = toplam + i
    return toplam
````

````python
print(topla(1,2,3,4))
````

````python
def kisi_bilgileri(bilgiler):
    print(f"Ad: {bilgiler['ad']}")
    print(f"Soyad: {bilgiler['soyad']}")

info = {
    "ad": "Ahmet",
    "soyad": "Yilmaz",
    "yas" : 22
}

info2 = {
    "ad": "Mehmet",
    "soyad": "Yilmaz",
    "yas" : 25,
    "meslek": "Mühendis"
}

kisi_bilgileri(info)
kisi_bilgileri(info2)
````

````python
def kisi_bilgileri(**kwargs):
    print(f"{kwargs} tip:{type(kwargs)}")

    for key, value in kwargs.items():
        print(f"{key} {value}")


kisi_bilgileri(ad="Ahmet", soyad="Yilmaz")
kisi_bilgileri(ad="Ahmet", soyad="Yilmaz", yas=22)
kisi_bilgileri(ad="Ahmet", soyad="Yilmaz", yas=22, meslek="Mühendis")
````

````python
def kisi_bilgileri(*args, **kwargs):
    print(f"{args} tip:{type(args)}")
    print(f"{kwargs} tip:{type(kwargs)}")

    for key, value in kwargs.items():
        print(f"{key} {value}")


kisi_bilgileri(1,2,3, ad="Ahmet", soyad="Yilmaz")
kisi_bilgileri(ad="Ahmet", soyad="Yilmaz", yas=22)
kisi_bilgileri(ad="Ahmet", soyad="Yilmaz", yas=22, meslek="Mühendis")
````
