```python
# ekleme modunda acildigi zaman imlec fd.seek ile hareket ettirilemez.
with open("ornek.text","a") as fd:
    fd.write("Yeni satir1")
    fd.seek(0)
    fd.write("Yeni satir2")
!cat ornek.text
```

```python
with open("ornek.text","w") as fd:
    for i in range(1,5):
        fd.write(f"{i}. Merhaba Dunya\n")


!cat ornek.text
```

```python
with open("ornek.text") as fd:
    for line in fd:
        print(line, end='')
```

```python
new_lines = ["5. Merhaba Dunya\n", "6. Merhaba Dunya\n"]
with open("ornek.text", "a") as dosya:
    dosya.writelines(new_lines)

!cat ornek.text
```
