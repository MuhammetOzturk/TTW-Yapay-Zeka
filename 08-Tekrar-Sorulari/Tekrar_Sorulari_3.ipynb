### 1. Sayı Kontrolü

* Sonucu ekrana yazdırın.


### 2. En Büyük Sayı Bulma

Kullanıcıdan üç farklı sayı alın.

**İstenenler:**

* Bu üç sayıdan **en büyüğünü** bulun ve ekrana yazdırın.


```python
sayi1 = int(input("Birinci sayıyı giriniz: "))
sayi2 = int(input("İkinci sayıyı giriniz: "))
sayi3 = int(input("Üçüncü sayıyı giriniz: "))

if sayi1 > sayi2 and sayi1 > sayi3:
    print("En büyük sayı: ", sayi1)
elif sayi2 > sayi1 and sayi2 > sayi3:
    print("En büyük sayı: ", sayi2)
else:
    print("En büyük sayı: ", sayi3)
```

```python
liste = [sayi1, sayi2, sayi3]
liste.sort()
liste[-1]
```

```python
max(liste)
```

```python
min(liste)
```

### 3. Haftanın Günü

Kullanıcıdan **1 ile 7 arasında** bir sayı alın.

**İstenenler:**

* Girilen sayıya karşılık gelen haftanın gününü yazdırın
  (1 → Pazartesi, 7 → Pazar).
* Geçersiz bir sayı girilirse **uyarı mesajı** verin.



### 4. Not Hesaplama (Çoklu Koşul)

Kullanıcıdan **0–100** aralığında bir sınav notu alın.

**Not Aralıkları:**

* 90–100 → AA
* 80–89 → BA
* 70–79 → BB
* 60–69 → CB
* 50–59 → CC
* 0–49  → FF

**İstenenler:**

* Notu kontrol edin ve harf notunu yazdırın.
* Geçersiz bir değer girilirse hata mesajı verin.
